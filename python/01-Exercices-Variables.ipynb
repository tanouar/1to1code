## Exercice 2 — Réutiliser et mettre à jour des variables

Rappel : Le signe `=` affecte une valeur à une variable. Une nouvelle affectation remplace la valeur précédente ; elle ne teste pas une égalité.

```python
prix_unitaire = 4.5
quantite = 3
total = prix_unitaire * quantite

print("Prix initial :", total, "euros")

quantite = 5
total = prix_unitaire * quantite
print("Après mise à jour de la quantité :", total, "euros")
```

```python
# Écrivez votre code ici
```

```python
# @title Correction
prix_billet = 12
nombre_billets = 3
montant_total = prix_billet * nombre_billets
print("Montant initial :", montant_total, "euros")

nombre_billets = 5
montant_total = prix_billet * nombre_billets
print("Montant mis à jour :", montant_total, "euros")
```

## Exercice 3 — Comprendre les booléens et les comparaisons

Rappel : Une comparaison comme `>=` produit un booléen : `True` si la comparaison est vraie, `False` sinon. `if` permet ensuite de choisir quel message afficher.

```python
age = 20
age_minimum = 18
est_majeur = age >= age_minimum

print("Est majeur :", est_majeur)
if est_majeur:
    print("Accès autorisé")
else:
    print("Accès refusé")
```

## Exercice 5 — Modifier une liste

Rappel : Contrairement à un tuple, une liste peut être modifiée après sa création. Vous pouvez remplacer une valeur avec son index, ajouter avec `append()` et retirer une valeur avec `remove()`.

```python
taches = ["lire", "réviser", "faire une pause"]
print("Départ :", taches)

taches[1] = "réviser Python"
taches.append("faire un exercice")
taches.remove("lire")

print("Après les modifications :", taches)
```

## Exercice 7 — Utiliser un dictionnaire

Rappel : Un dictionnaire associe des clés à des valeurs. On retrouve une information grâce à sa clé plutôt que grâce à sa position.

```python
livre = {
    "titre": "Le Petit Prince",
    "auteur": "Antoine de Saint-Exupéry",
    "pages": 96
}

print("Titre :", livre["titre"])
livre["pages"] = 100
livre["langue"] = "français"
del livre["auteur"]
print("Dictionnaire après modifications :", livre)
```

**À vous de jouer :** créez un dictionnaire `film` avec les clés `titre`, `annee` et `note`. Affichez l'année, modifiez la note, ajoutez la clé `genre`, puis affichez le dictionnaire.

```python
# Écrivez votre code ici
```

```python
# @title Correction
film = {
    "titre": "Le Voyage de Chihiro",
    "annee": 2001,
    "note": 8.5
}

print("Année :", film["annee"])
film["note"] = 9
film["genre"] = "animation"
print(film)
```

## Exercice 8 — Choisir le bon type de données

Rappel : Le bon type dépend de l'usage : une liste convient à une collection modifiable, un tuple à des valeurs fixes et un dictionnaire à des informations repérées par une clé.

```python
profil = {
    "nom": "Samira",
    "competences": ["Python", "SQL"],
    "position": (48.86, 2.35),
    "disponible": True
}

print("Nom :", profil["nom"])
print("Première compétence :", profil["competences"][0])
print("Latitude :", profil["position"][0])
print("Disponible :", profil["disponible"])
```
