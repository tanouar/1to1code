# Mécanisme d'assignation

En Python, vous pouvez créer une variable en lui assignant une valeur à l'aide de l'opérateur `=`. Il ne faut pas confondre l'opérateur d'assignation `=` avec le signe mathématique **=**. 

L'opérateur `=` est utilisé pour assigner une valeur à une variable.

Cela signifie en language machine que "quelque chose" est stocké en mémoire et que nous lui donnons un nom pour pouvoir y accéder plus tard. Pour cela Python est très pratique car on ne s'occupe pas de savoir ou se trouve nos objets en mémoire, on peut juste les utiliser grâce à leur nom.

Representez vous l'assignation comme une étiquette que vous collez sur une boîte pour pouvoir la retrouver facilement plus tard. Et a chaque fois qu'on rapelle cette boite on peut y acceder, la transoformer, la manipuler, etc...

# int, float, str

Les types de variables les plus courants en Python sont :

- `int` : pour les entiers (ex : 5, -10)
- `float` : pour les nombres à virgule flottante (ex : 3.14, -0.001)
- `str` : pour les chaînes de caractères (ex : "Bonjour", 'Python')

Python est assez intelligent pour inférer le type d'une variable en fonction de la valeur qui lui est assignée. Par exemple, si vous assignez un nombre entier à une variable, Python la considérera comme un `int`. Si vous assignez une chaîne de caractères, elle sera considérée comme un `str`.

par exemple si vous faites :

````python
x = 5
y = 3.14
text = "Bonjour"
````

vous créez trois variables : `x` est un `int`, `y` est un `float`, et `text` est un `str`.
On remarque que pour une chaîne de caractères ou str, il est nécessaire/obligatoire d'utiliser des guillemets simples ou doubles. de préférence on utilisera des guillemets doubles pour les chaînes de caractères.

# Les booléens : bool

Les booléens en Python représentent une valeur de vérité : soit `True`, soit `False`. Ils sont souvent utilisés dans les conditions et les boucles pour contrôler le flux d'exécution d'un programme. Par exemple :

````python
est_vrai = True
est_faux = False
````

Vous pouvez également obtenir des valeurs booléennes à partir d'expressions :

````python
x = 5
y = 10
print(x == y)  # False
````

> Ici il est important de noter que `==` est un opérateur de comparaison qui vérifie si les valeurs de `x` et `y` sont strictement identiques, tandis que `=` est un opérateur d'affectation qui assigne une valeur à une variable.

Les booléens sont souvent utilisés dans des instructions conditionnelles :

````python
ceci = True 

if ceci:
    print("C'est vrai !")
else:
    print("C'est faux !")
````

Retenez également que `True` vaut **1** et `False` vaut **0**, il est important de garder cela à l'esprit lorsque nous utiliserons des fonctions qui retournent des booléens dans les cours suivants.

**Point à retenir :** un booléen sera toujours soit `True`, soit `False`.

**Remarque** :  Les booléens sont essentiels pour la logique de programmation, car ils permettent de prendre des décisions basées sur des conditions. Ils sont également utilisés dans les opérations logiques comme `and`, `or`, et `not` pour combiner ou inverser des conditions. 

Si vous débutez en programmation, il est normal de ne pas tout comprendre l'intérêt de ce type de variable, toutefois retenez que les booléens sont omniprésents dans la programmation et que vous les utiliserez très souvent pour contrôler le flux de votre code. Ne vous inquiétez pas si cela semble abstrait au début, avec la pratique, vous verrez à quel point les booléens sont puissants et utiles !

# Les listes Python : list 

Les list sont très utiles en Python, elles permettent de stocker plusieurs valeurs dans une seule variable. Par exemple :

````python
nombres = [1, 2, 3, 4, 5]
````

Ici, `nombres` est une liste d'entiers. Vous pouvez accéder à chaque élément de la liste en utilisant son index (qui commence à 0) :

````python
print(nombres[0])  # Affiche 1
print(nombres[1])  # Affiche 2
````

Les listes peuvent contenir des éléments de différents types :

````python
melange = [1, "deux", 3.0, True]
````

Vous pouvez également modifier les éléments d'une liste :

````python
nombres[0] = 10
print(nombres)  # Affiche [10, 2, 3, 4, 5]
````

Les listes en Python sont dynamiques, ce qui signifie que vous pouvez ajouter ou supprimer des éléments à tout moment :

````python
nombres.append(6)  # Ajoute 6 à la fin de la liste
print(nombres)  # Affiche [10, 2, 3, 4, 5, 6]
nombres.remove(2)  # Supprime 2 de la liste
print(nombres)  # Affiche [10, 3, 4, 5, 6]
````

**Point à retenir :** une liste sera toujours contenue entre crochets `[]`.

*Remarque : il n'est pas possible de nommer une liste avec un nom de type natif comme `list`, `int`, `str`, etc. Cela peut causer des problèmes dans votre code, car vous ne pourrez plus utiliser ces types natifs pour créer de nouvelles variables.*

# Les tuples Python : tuple

Les tuples sont similaires aux listes, mais ils sont immuables, ce qui signifie que vous ne pouvez pas modifier leurs éléments une fois qu'ils ont été créés. Ils sont définis en utilisant des parenthèses `()` au lieu de crochets `[]`. Par exemple :

````python
coordonnees = (10.0, 20.0)
````

Ici, `coordonnees` est un tuple contenant deux valeurs flottantes. Vous pouvez accéder aux éléments d'un tuple de la même manière qu'avec une liste :

````python
print(coordonnees[0])  # Affiche 10.0
print(coordonnees[1])  # Affiche 20.0
````

Si vous essayez de modifier un élément d'un tuple, vous obtiendrez une erreur :

````python
coordonnees[0] = 15.0  # Erreur : TypeError
````

Les tuples sont souvent utilisés pour représenter des données qui ne doivent pas être modifiées, comme les coordonnées d'un point ou les valeurs d'un enregistrement.

**Point à retenir :** un tuple sera toujours contenu entre parenthèses `()`.

> Les tuples sont immuables, ce qui les rend plus sûrs pour stocker des données qui ne doivent pas être modifiées. Ils sont également plus rapides que les listes pour certaines opérations, car ils sont optimisés pour l'immuabilité. Utilisez des tuples lorsque vous avez besoin de stocker des données qui ne doivent pas changer, et des listes lorsque vous avez besoin de données modifiables.

# Les dictionnaires Python : dict

Les dictionnaires en Python sont des collections de paires clé-valeur. Ils sont très utiles pour stocker des données associatives. Par exemple :

````python
personne = {
    "nom": "Alice",
    "age": 30,
    "ville": "Paris"
}
````

Ici, `personne` est un dictionnaire avec trois paires clé-valeur. Vous pouvez accéder à une valeur en utilisant sa clé :

````python
print(personne["nom"])  # Affiche "Alice"
print(personne["age"])  # Affiche 30
````

Vous pouvez également ajouter ou modifier des éléments dans un dictionnaire :

````python
personne["age"] = 31  # Modifie l'âge
personne["profession"] = "Ingénieur"  # Ajoute une nouvelle clé
print(personne)
````

Pour supprimer un élément, vous pouvez utiliser le mot-clé `del` :

````python
del personne["ville"]
print(personne)
````

**Point à retenir :** un dictionnaire sera toujours contenu entre accolades `{}`.

> Le type de données `dict` est très puissant et offre de nombreuses **méthodes** (une méthode en python, sont les fonctions permettant de manipuler nos objets, nous y reviendrons plus tard) pour manipuler les données, comme `keys()`, `values()`, `items()`, etc. Il est largement utilisé dans la programmation Python pour organiser et gérer des données complexes. Ne negligez pas l'apprentissage des dictionnaires, car ils sont partout dans la programmation Python et en data science !

# En résumé

| Type | Syntaxe | Mutable ? | Exemple | Cas d'usage typique |
|---|---|---|---|---|
| `int` | `x = 42` | ✅ | `age = 30` | Compteurs, indices, quantités |
| `float` | `x = 3.14` | ✅ | `prix = 9.99` | Mesures, calculs scientifiques |
| `str` | `x = "texte"` | ❌ | `nom = "Alice"` | Noms, messages, identifiants |
| `bool` | `x = True` | ✅ | `actif = False` | Conditions, drapeaux logiques |
| `list` | `x = [...]` | ✅ | `[1, 2, 3]` | Collections ordonnées modifiables |
| `tuple` | `x = (...)` | ❌ | `(10.0, 20.0)` | Données fixes, coordonnées |
| `dict` | `x = {...}` | ✅ | `{"nom": "Alice"}` | Données associatives, configurations |

> **Règle d'or** : choisissez le type le plus adapté à vos données. Si une donnée ne doit jamais changer, préférez un `tuple` à une `list`. Si vous avez besoin d'associer des clés à des valeurs, un `dict` est toujours plus lisible qu'une `list` de `list`.
