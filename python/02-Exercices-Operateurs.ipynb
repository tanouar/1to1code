## Exercice 4 - Vérifier une quantité

**Rappel** : une comparaison produit ici `True` ou `False`. `=` conserve une valeur, tandis que `==` teste une égalité. La démonstration vérifie une commande par rapport au stock.

````python
stock = 12
quantite_demandee = 10
print(stock >= quantite_demandee)
print(stock == quantite_demandee)
````

**À vous de jouer** :
1. Initialisez `stock = 12` et `quantite_demandee = 15`.
2. Vérifiez si le stock suffit avec `>=`.
3. Vérifiez si la quantité est strictement positive et inférieure ou égale au stock, avec une comparaison chaînée.
4. Vérifiez si les deux quantités sont différentes avec `!=`.

**Résultat attendu** :
```output
Stock suffisant : False
Quantité valide : False
Quantités différentes : True
```
**Question de réflexion** : que donnerait la comparaison chaînée pour une quantité de 0 ?

````python
# À vous de jouer : écrivez votre code ici.
````

````python
# @title Correction
stock = 12
quantite_demandee = 15
print("Stock suffisant :", stock >= quantite_demandee)
print("Quantité valide :", 0 < quantite_demandee <= stock)
print("Quantités différentes :", stock != quantite_demandee)
````

## Exercice 7 - Appartenance, égalité et identité

**Rappel** : `in` cherche un élément ; `==` compare les valeurs ; `is` compare l'identité des objets. `.copy()` crée ici une autre liste contenant les mêmes textes.

````python
formats = ["csv", "json"]
copie = formats.copy()
print("csv" in formats)
print(formats == copie)
print(formats is copie)
````

**À vous de jouer** :
1. Créez `original = ["csv", "json"]`, `autre_nom = original` et `copie = original.copy()`.
2. Vérifiez l'absence de `"xml"` avec `not in`.
3. Comparez `original` et `copie` avec `==`, puis avec `is`.
4. Vérifiez si `original` et `autre_nom` désignent le même objet.

**Résultat attendu** :
```output
XML absent : True
Même contenu : True
Même objet que la copie : False
Même objet que autre_nom : True
```
**Question de réflexion** : si vous ajoutez un élément à `autre_nom`, quelles listes changent ? Testez après avoir obtenu le résultat demandé.

````python
# À vous de jouer : écrivez votre code ici.
````

````python
# @title Correction
original = ["csv", "json"]
autre_nom = original
copie = original.copy()
print("XML absent :", "xml" not in original)
print("Même contenu :", original == copie)
print("Même objet que la copie :", original is copie)
print("Même objet que autre_nom :", original is autre_nom)
````

## Exercice 8 - Synthèse : calculer les frais de livraison

**Rappel** : Python exécute la première branche vraie d'une chaîne `if` / `elif` / `else`. La démonstration choisit un message à partir d'une quantité.

````python
quantite = 3
if quantite >= 10:
    print("Grand lot")
elif quantite > 0:
    print("Petit lot")
else:
    print("Aucun article")
````

**À vous de jouer** : reproduisez la règle du cours.
1. Initialisez `montant = 75` et `client_fidele = True`.
2. Affectez 0 à `frais` si le montant est au moins 100.
3. Sinon, affectez 3 si le client est fidèle, et 6 dans les autres cas.
4. Calculez `total = montant + frais` et affichez les deux montants.

**Résultat attendu** :
```output
Frais : 3
Total : 78
```
**Question de réflexion** : testez ensuite un montant de 100 avec un client non fidèle, puis 75 avec un client non fidèle. Vous devez obtenir respectivement 0 et 6 de frais. Pourquoi tester exactement la borne de 100 ?

````python
# À vous de jouer : écrivez votre code ici.
````

````python
# @title Correction
montant = 75
client_fidele = True
if montant >= 100:
    frais = 0
elif client_fidele:
    frais = 3
else:
    frais = 6
total = montant + frais
print("Frais :", frais)
print("Total :", total)
````
