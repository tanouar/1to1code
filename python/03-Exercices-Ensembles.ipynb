## Exercice 3 - Chercher et supprimer

**Rappel** : `in` vérifie la présence. `remove()` signale une erreur si l'élément manque ; `discard()` accepte son absence. Dans la démonstration, l'absence de `"xml"` est prévue.

````python
formats = {"csv", "json"}
print("csv" in formats)
formats.remove("csv")
formats.discard("xml")
print(sorted(formats))
````

**Question de réflexion** : quelle erreur obtiendriez-vous avec `remove("David")`, et dans quel cas serait-elle utile ?

## Exercice 4 - Réunir deux groupes

**Rappel** : l'union `a | b` réunit tous les éléments sans doublon et sans modifier les ensembles de départ. La démonstration compare l'union et la mise à jour en place.

````python
a = {1, 2}
b = {2, 3}
reunion = a.union(b)
print(sorted(reunion))
print(sorted(a))
a.update(b)
print(sorted(a))
````

## Exercice 6 - Distinguer les différences

**Rappel** : `a - b` conserve les éléments de A absents de B ; `a ^ b` conserve ceux présents dans un seul des deux ensembles. La démonstration montre que le sens de la soustraction compte.

````python
a = {1, 2, 3}
b = {3, 4}
print(sorted(a - b))
print(sorted(b - a))
print(sorted(a ^ b))
````

**À vous de jouer** :

**Question de réflexion** : pourquoi Bob est-il absent des trois résultats ?

## Exercice 7 - Comparer trois collections

**Rappel** : `union()`, `intersection()` et `difference()` acceptent plusieurs collections. Une intersection conserve ce qui est commun à toutes. La démonstration travaille avec trois groupes de nombres.

````python
a = {1, 2, 3}
b = {2, 3, 4}
c = {3, 5}
print(sorted(a.union(b, c)))
print(sorted(a.intersection(b, c)))
print(sorted(a.difference(b, c)))
````

## Exercice 8 - Synthèse : préparer une relance

**Rappel** : une différence identifie les absents ; une inclusion vérifie que tous les éléments d'un groupe appartiennent à un autre. La démonstration contrôle des formats nécessaires.

````python
necessaires = {"csv", "json"}
disponibles = {"csv", "json", "parquet"}
print(necessaires <= disponibles)
print(necessaires.isdisjoint({"xml"}))
````

**À vous de jouer** : préparez le bilan d'un atelier.
1. Créez la liste `inscriptions = ["Alice", "Bob", "Alice", "Clara", "David"]`, puis l'ensemble `inscrits`.
2. Créez `presents = {"Alice", "Clara"}`.
3. Calculez `a_relancer = inscrits - presents`.
4. Affichez le nombre d'inscrits distincts, les personnes à relancer, puis vérifiez que tous les présents sont inscrits et qu'aucun présent n'est à relancer.

**Résultat attendu** :
```output

À relancer : ['Bob', 'David']
Présents tous inscrits : True
Groupes disjoints : True
```
**Question de réflexion** : si vous ajoutez une personne non inscrite à `presents`, quel contrôle passe à `False` ?

````python
# À vous de jouer : écrivez votre code ici.
````

````python
# @title Correction
inscriptions = ["Alice", "Bob", "Alice", "Clara", "David"]
inscrits = set(inscriptions)
presents = {"Alice", "Clara"}
a_relancer = inscrits - presents
print("Inscrits distincts :", len(inscrits))
print("À relancer :", sorted(a_relancer))
print("Présents tous inscrits :", presents <= inscrits)
print("Groupes disjoints :", presents.isdisjoint(a_relancer))
````
