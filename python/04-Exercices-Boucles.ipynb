## Exercice 3 - Numéroter des éléments

**Rappel** : `enumerate()` fournit un numéro et une valeur. Par défaut, la numérotation commence à 0, comme dans la démonstration.

````python
formats = ["csv", "json"]
for index, format_fichier in enumerate(formats):
    print(index, format_fichier)
````

````python
# À vous de jouer : écrivez votre code ici.
````

````python
# @title Correction
etapes = ["lecture", "validation", "enregistrement"]
for numero, etape in enumerate(etapes, start=1):
    print(numero, etape)
````

## Exercice 4 - Parcourir un dictionnaire et additionner

**Rappel** : `.items()` fournit les paires clé-valeur. Un accumulateur est initialisé avant la boucle puis mis à jour avec `+=`. La démonstration compte des fichiers par format.

````python
fichiers = {"csv": 3, "json": 2}
total = 0
for format_fichier, quantite in fichiers.items():
    print(format_fichier, quantite)
    total += quantite
print("Total :", total)
````

**À vous de jouer** :
1. Créez `stocks = {"carnets": 4, "stylos": 10}` et `total = 0`.
2. Parcourez les produits et les quantités avec `.items()`.
3. Affichez chaque produit avec sa quantité, puis ajoutez la quantité au total.
4. Affichez le total après la boucle.

**Résultat attendu** :
```output
carnets : 4
stylos : 10
Total : 14
```
**Question de réflexion** : quel total obtiendriez-vous si `total = 0` était placé au début du corps de la boucle ?

````python
# À vous de jouer : écrivez votre code ici.
````

````python
# @title Correction
stocks = {"carnets": 4, "stylos": 10}
total = 0
for produit, quantite in stocks.items():
    print(produit, ":", quantite)
    total += quantite
print("Total :", total)
````

## Exercice 5 - Construire un compte à rebours

**Rappel** : `while` teste la condition avant chaque passage. Le compteur doit évoluer pour que la boucle s'arrête. La démonstration part de 3.

````python
compteur = 3
while compteur > 0:
    print(compteur)
    compteur -= 1
print("Départ !")
````

**À vous de jouer** :
1. Initialisez `compteur = 4`.
2. Tant que le compteur est strictement positif, affichez-le puis diminuez-le de 1.
3. Affichez `"Terminé"` après la boucle.

**Résultat attendu** :
```output
4
3
2
1
Terminé
```
**Question de réflexion** : si vous commencez à 0, combien de fois le corps s'exécute-t-il ? Ne supprimez pas la diminution du compteur : elle garantit l'arrêt.

````python
# À vous de jouer : vérifiez la mise à jour du compteur avant d'exécuter.
````

````python
# @title Correction
compteur = 4
while compteur > 0:
    print(compteur)
    compteur -= 1
print("Terminé")
````

## Exercice 6 - S'arrêter avec break

**Rappel** : `break` quitte la boucle en cours. La démonstration arrête une recherche dès que le nom attendu est rencontré.

````python
fichiers = ["notes.txt", "ventes.csv", "clients.csv"]
for fichier in fichiers:
    if fichier == "ventes.csv":
        print("Trouvé :", fichier)
        break
    print("Autre fichier :", fichier)
````

**À vous de jouer** :
1. Créez `valeurs = [4, 7, 12, 15]`.
2. Parcourez les valeurs : si une valeur est supérieure ou égale à 10, affichez `"Seuil atteint :"` et cette valeur, puis arrêtez la boucle.
3. Sinon, affichez `"Sous le seuil :"` et la valeur.
4. Affichez `"Recherche terminée"` après la boucle.

**Résultat attendu** :
```output
Sous le seuil : 4
Sous le seuil : 7
Seuil atteint : 12
Recherche terminée
```
**Question de réflexion** : pourquoi 15 n'est-il jamais affiché ? Que se passe-t-il avec `[4, 7]` ?

````python
# À vous de jouer : écrivez votre code ici.
````

````python
# @title Correction
valeurs = [4, 7, 12, 15]
for valeur in valeurs:
    if valeur >= 10:
        print("Seuil atteint :", valeur)
        break
    print("Sous le seuil :", valeur)
print("Recherche terminée")
````

## Exercice 7 - Ignorer un passage avec continue

**Rappel** : `continue` ignore la fin du bloc pour l'élément actuel, mais poursuit le parcours. La démonstration ignore les mesures négatives.

````python
mesures = [12, -1, 8]
for mesure in mesures:
    if mesure < 0:
        continue
    print(mesure)
````

**À vous de jouer** :
1. Créez `mesures = [-1, 5, 0, -3, 8]`.
2. Dans une boucle `for`, utilisez `continue` si une mesure est négative.
3. Affichez les autres mesures, y compris zéro.

**Résultat attendu** :
```output
5
0
8
```
**Question de réflexion** : quel serait l'affichage si vous remplaciez `continue` par `break` ?

````python
# À vous de jouer : écrivez votre code ici.
````

````python
# @title Correction
mesures = [-1, 5, 0, -3, 8]
for mesure in mesures:
    if mesure < 0:
        continue
    print(mesure)
````

## Exercice 8 - Synthèse : produire un bilan de mesures

**Rappel** : on peut combiner une condition, une liste résultat et un accumulateur. La démonstration copie les mesures valides dans une nouvelle liste pour ne pas modifier celle parcourue.

````python
mesures = [3, -1, 2]
valides = []
for mesure in mesures:
    if mesure >= 0:
        valides.append(mesure)
print(valides)
````

**À vous de jouer** :
1. Créez `mesures = [12, -1, 8, 0, 5]`, `valides = []` et `total = 0`.
2. Parcourez les mesures et ignorez les valeurs négatives avec `continue`.
3. Pour chaque valeur conservée, ajoutez-la à `valides` et au `total`.
4. Après la boucle, affichez la liste, sa longueur avec `len()` et le total.

**Résultat attendu** :
```output
Mesures valides : [12, 8, 0, 5]
Nombre : 4
Total : 25
```
**Question de réflexion** : vérifiez aussi les cas `[]` et `[-2, -1]`. Pourquoi le bilan reste-t-il défini même si aucune mesure n'est conservée ?

````python
# À vous de jouer : écrivez votre code ici.
````

````python
# @title Correction
mesures = [12, -1, 8, 0, 5]
valides = []
total = 0
for mesure in mesures:
    if mesure < 0:
        continue
    valides.append(mesure)
    total += mesure
print("Mesures valides :", valides)
print("Nombre :", len(valides))
print("Total :", total)
````
