## Exercice 7 - Documenter une moyenne

**Rappel** : une docstring explique le contrat d'une fonction. `raise ValueError(...)` signale une valeur d'entrée inadaptée. La démonstration exige une liste non vide plutôt que de retourner une fausse moyenne de zéro.

````python
def calculer_moyenne(notes):
    """Renvoie la moyenne ; une liste vide déclenche une ValueError."""
    if not notes:
        raise ValueError("La liste de notes ne doit pas être vide.")
    return sum(notes) / len(notes)

print(calculer_moyenne([12, 15, 18]))
````

**À vous de jouer** :

2. Si la liste est vide, utilisez `raise ValueError("La liste de notes ne doit pas être vide.")`.
3. Sinon, renvoyez la moyenne arrondie à deux décimales avec `round(sum(notes) / len(notes), 2)`.
4. Affichez le résultat pour `[10, 11, 13]`.

**Résultat attendu** :
```output

```
**Question de réflexion** : consultez ensuite `help(moyenne_arrondie)` dans une cellule séparée. Quelle information permet d'anticiper le comportement sur `[]` ? Un appel avec `[]` doit déclencher une `ValueError`, pas renvoyer 0.

````python
# À vous de jouer : écrivez votre fonction puis son appel.
````

````python
# @title Correction
def moyenne_arrondie(notes):
    """Renvoie la moyenne d'une liste non vide de nombres, arrondie à deux décimales.

    Une liste vide déclenche une ValueError.
    """
    if not notes:
        raise ValueError("La liste de notes ne doit pas être vide.")
    return round(sum(notes) / len(notes), 2)

print(moyenne_arrondie([10, 11, 13]))
````

## Exercice 10 - Synthèse : réutiliser un bilan

**Rappel** : une fonction peut contenir une boucle et renvoyer les résultats accumulés. Nous allons reprendre le filtrage des mesures du chapitre précédent. La démonstration compte les valeurs non négatives.

````python
def compter_valides(mesures):
    nombre = 0
    for mesure in mesures:
        if mesure >= 0:
            nombre += 1
    return nombre

print(compter_valides([3, -1, 0]))
````

**À vous de jouer** :
1. Définissez `resumer_mesures(mesures)` et ajoutez une docstring : elle renvoie le nombre et la somme des mesures non négatives, sans modifier la liste reçue.
2. Initialisez `nombre` et `total` à 0 dans la fonction.
3. Parcourez les mesures. Pour chaque valeur supérieure ou égale à 0, augmentez `nombre` de 1 et ajoutez la valeur au total.
4. Renvoyez le tuple `(nombre, total)` après la boucle.
5. Affichez les retours pour `[12, -1, 8, 0, 5]`, puis pour `[]`.

**Résultat attendu** :
```output

(0, 0)
```
**Question de réflexion** : pourquoi la liste vide peut-elle produire un nombre et une somme de zéro ici, alors qu'une moyenne sur une liste vide doit être signalée comme invalide ?

````python
# À vous de jouer : écrivez votre fonction puis ses appels.
````

````python
# @title Correction
def resumer_mesures(mesures):
    """Renvoie le nombre et la somme des mesures non négatives sans modifier la liste.

    Une liste vide renvoie (0, 0).
    """
    nombre = 0
    total = 0
    for mesure in mesures:
        if mesure >= 0:
            nombre += 1
            total += mesure
    return nombre, total

print(resumer_mesures([12, -1, 8, 0, 5]))
print(resumer_mesures([]))
````
